Compare math scores for male vs female students

```python
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats
import pandas as pd
```

```python
students=pd.read_csv("StudentsPerformance.csv")
students.head()
```

```python
# calculate the mean of the males and females in the math score
male_mean=students[students['gender']=='male']['math score'].mean()
female_mean=students[students['gender']=='female']['math score'].mean()
print(male_mean)
print(female_mean)
```

```python
alpha=0.05
```

```python
# perfom a t_test
t_stat, p_val = stats.ttest_ind(
    students[students['gender'] == 'male']['math score'],
    students[students['gender'] == 'female']['math score']
)
print(t_stat)
print(p_val)
```

```python
Ho="There is no difference in the population and sample mean of the maths test"
Ha="There is a significant difference in the population and sample mean of the maths test"
if p_val< alpha :
    print("Reject Ho")
else:
    print("Fail to reject  Ho")
```

Paired Samples t-Test

Compare the mean math and mean reading scores of the same students.

```python
# Calculate both means
math_mean = students['math score'].mean()
reading_mean = students['reading score'].mean()
print(math_mean)
print(reading_mean)
```

```python
# Perform paired t-test
t_stat, p_val = stats.ttest_rel(students['math score'], students['reading score'])
print(f"T-statistic: {t_stat:.3f}")
print(f"P-value: {p_val:.4f}")
```

```python
alpha=0.05
```

```python
Ho="There is no difference in the population and sample mean of the maths test"
Ha="There is a significant difference in the population and sample mean of the maths test"
if p_val< alpha :
    print("Reject Ho")
else:
    print("Fail to reject  Ho")
```
