```python
import pandas as pd
import seaborn as sns  
from scipy import stats 
import matplotlib.pyplot as plt 
```

ONE SAMPLE T-TEST

```python
# load the dataset 
students=pd.read_csv("StudentsPerformance.csv")
students.head()
```

```python
# renam the columns
students.rename(columns={'math score':"math",'reading score':'reading','writing score':'writing'},inplace=True)
students.head()
```

```python
# checking for null vallues in the maths column
students.isnull().sum()
```

```python
population_mean=70
alpha=0.05
# calculate sample mean which is the maths scores
sample_mean=students['math'].mean()
print(sample_mean)

```

```python
t_stat,p_val=stats.ttest_1samp(students['math'],population_mean)
print("t:",t_stat
      ,"p:",p_val)
```

```python
Ho="There is no difference in the population and sample mean of the maths test"
Ha="There is a significant difference in the population and sample mean of the maths test"
if p_val< alpha :
    print("Reject Ho")
else:
    print("Fail to reject  Ho")
```

```python
sns.histplot(students["math"], kde=True, color='lightblue')
plt.axvline(sample_mean, color='red', linestyle='--', label=f"Sample Mean = {sample_mean:.2f}")
plt.axvline(population_mean, color='green', linestyle='-', label="Population Mean = 70")
plt.title("One-Sample t-Test: Math Scores vs 70")
plt.legend()
plt.show()
```
